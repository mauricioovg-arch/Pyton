## 9. Clases y métodos

Una clase es una plantilla para crear objetos

Por ejemplo: una cuenta bancaria puede tener:

- Estado: titular, saldo.
- Métodos: depositar, retirar, consultar

````python
class Cuenta:
 def __init__(self, titular, saldo=0):
  self.titular = titular
  self.saldo = saldo
 def depositar(self, monto):
     if monto <= 0:
         return "Monto invalido"
     self.saldo += monto
     return f"Depositaste {monto}. Tu saldo actual es {self.saldo}"
 def retirar(self, monto):
     if monto <= 0 or monto > self.saldo:
         return "NO DIGAS MMDS"
     self.saldo -= monto
     return f"Retiraste {monto}. Tu sldo actual es {self.saldo}"
 def consultar_saldo(self):
     return f"Tu saldo actual es {self.saldo}"

wincuenta = Cuenta("Wino", 100)
print(wincuenta.consultar_saldo())
print(wincuenta.depositar(50))
print(wincuenta.retirar(3020))
print(wincuenta.consultar_saldo())
 
    
````

## 10. Herencia

La herencia permite crear una clase nueva a partir de una existente.

La clase nueva puede:

- Reutilizar métodos
- Agregar atributos nuevos
- Agregar métodos nuevos
- Modificar mérodos existentes

|Concepto|	Qué es|
|-|-|
|Clase Padre|	La clase original con métodos y atributos|
|Clase Hija	|Hereda todo de la clase padre|
|Herencia	|Compartir código entre clases|
|`super()`	|Llamar métodos de la clase padre|
|Override	|Sobrescribir un método heredado|

````python
#Aqui usaremos la clase hecha en el inscio anterior

class CuentaAhorro(Cuenta):
    def __init__(self, titular, saldo=0, meta=1999):
        super().__init__(titular, saldo)
        self.meta = meta
    def avamzar_meta(self, monto):
        self.depositar(monto)
        
        if self.saldo >= self.meta:
            return f"¡Felicidades! meta alcanzada. Tu saldo:{self.saldo}, Meta:{self.meta}"
        else:
            return f"Avance:{self.saldo} de {self.meta}. Continua asi"
        
ahorro = CuentaAhorro("Wina", 1000, 2000)
print(ahorro.avamzar_meta(500))

````

### Ahora hagamos un ejercicio:
Vamos a crear una clase llamada *Producto* con:
Atributos:
- nombre
- precio
- cantidad
  
Métodos:
- vender (cantidad): disminuira la cantidad si hay suficiente inventario.
- reabastecer (cantidad): aumenta la cantidad.
- mostrar(): muestra el estado del producto.
  
Después crear una clase *ProductoPerecedero* que herede de *Producto* y agregue:
- fecha_caducidad
- esta_vigente(): regresa *True* si la fecha no está vacia, como ejemplo simple.
  
Se hará en el siguiente recuadro:

````python
class Producto:
    def __init__(self, nombre, precio=0, cantidad=0):
        self.nombre = nombre
        self.precio = precio
        self.cantidad = cantidad
    def vender(self, cantidad_vendida):
        if cantidad_vendida > self.cantidad:
            print("No hay suficiente inventario para vender esa cantidad.")
        elif cantidad_vendida == self.cantidad:
            self.cantidad -= cantidad_vendida
            print(f"Se vendieron {cantidad_vendida} unidades de {self.nombre} \n Ya no quedan articulos de {self.nombre}")
        elif cantidad_vendida < self.cantidad:
            self.cantidad -= cantidad_vendida
            print(f"Se vendieron {cantidad_vendida} unidades de {self.nombre} \n Quedan {self.cantidad} de {self.nombre}")
        else:
            print("Cantidad inválida.")  
                    
    def reabastecer(self, cantidad_reabastecida):
        if cantidad_reabastecida <= 0:
            print("No te hagas pendejo, ponte a reabastecer")
        else:
            self.cantidad += cantidad_reabastecida 
            print(f"Acabas de reabastecer {cantidad_reabastecida} de {self.nombre}")
    def mostrar_inventario(self):
        return f"Producto: {self.nombre}, Precio: {self.precio}, Cantidad: {self.cantidad}"
#Otra clase
class ProductoPerecedero(Producto):
    def __init__(self, nombre, fecha_caducidad, precio=0, cantidad=0, vigente=True):
        super().__init__(nombre, precio, cantidad)
        self.fecha_caducidad = fecha_caducidad
        self.vigente = vigente
    def LeerCaducidad(self, fecha_caducidad):
        print(f"La fecha de caduciad de {self.nombre} es {self.fecha_caducidad}")
    def Vigencia(self):
       if self.vigente == True:
           print("Su producto esta vigente")
       else:
           print("No esta vigente, tira esa chingadera")

# Vamos a usar el codigo
Winopardo = ProductoPerecedero("Chupas", "20.11.19", 300, 20, True)
print(Winopardo.fecha_caducidad)
Winopardo.reabastecer(20)
Winopardo.Vigencia()
````
